```python
import math

def cosine_similarity(a, b):
    dot_product = sum(x * y for x, y in zip(a, b))
    length_a = math.sqrt(sum(x * x for x in a))
    length_b = math.sqrt(sum(y * y for y in b))

    if length_a == 0 or length_b == 0:
        return 0

    return dot_product / (length_a * length_b)


# Example vectors for learning
question_vector = [1, 2, 3]
similar_vector = [2, 4, 6]
different_vector = [3, 1, 0]

print("Similar:", round(cosine_similarity(
    question_vector, similar_vector
), 2))

print("Different:", round(cosine_similarity(
    question_vector, different_vector
), 2))
```

```python
import re

def answer_question_ai(question):
    results = search_notes(question)

    if not results:
        print("I couldn't find relevant information in your notes.")
        return

    context = "\n\n".join(
        item["text"] for score, item in results[:2]
    )

    response = client.chat.completions.create(
        model="spur-qwen3-8b",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a friendly Grade 7 homework helper. "
                    "Answer simply and accurately using only the "
                    "provided study notes. If the notes do not "
                    "contain the answer, say so. Give only the "
                    "final answer. Never include internal analysis "
                    "or <think> tags."
                )
            },
            {
                "role": "user",
                "content": (
                    f"Study notes:\n{context}\n\n"
                    f"Question: {question}\n"
                    "Give a short answer."
                )
            }
        ],
        max_tokens=150
    )

    answer = response.choices[0].message.content or ""
    answer = re.sub(
        r"<think>.*?</think>",
        "",
        answer,
        flags=re.DOTALL
    ).strip()

    print("Question:", question)
    print("\nAnswer:", answer)
    print("\nSource:", results[0][1]["source"])
```

```python
answer_question_ai("Why does the Moon shine?")
```
